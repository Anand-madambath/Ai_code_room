```python
os.environ['GEMINI_API_KEY'] = userdata.get("gemini")
client = genai.Client()

history = []
config = types.GenerateContentConfig(
    system_instruction="you are a friendly tutor. keep answers short",
    max_output_tokens=2000,
    temperature=0.7,
)

while True:
    user_text = input("you: ")
    if user_text == "quit":
        break

    if user_text.strip() == "":
        print("Please type something.")
        continue

    if user_text == "/reset":
        history = []
        print("history reset")
        continue

    history.append(types.Content(role="user", parts=[types.Part(text=user_text)]))

    full_text = ""
    last_usage = None
    try:
        stream = client.models.generate_content_stream(
            model="gemini-2.5-flash-lite",
            contents=history,
            config=config,
        )
        print("Bot: ", end="")
        for chunk in stream:
            if chunk.usage_metadata is not None:
                last_usage = chunk.usage_metadata
            if chunk.text:
                print(chunk.text, end="", flush=True)
                full_text += chunk.text
        print()
    except Exception as e:
        print("Api error:", e)
        history.pop()
        continue

    if not full_text:
        print("Model returned nothing.")
        history.pop()
        continue

    history.append(types.Content(role="model", parts=[types.Part(text=full_text)]))

    if last_usage is not None:
        print("input tokens:", last_usage.prompt_token_count)
```

```python
prompt = "Give me a name for a coffee shop in Thrissur. Reply with only the name."

for temp in [0.0, 1.5]:
    print(f"--- temperature = {temp} ---")
    for run in range(2):
        resp = client.models.generate_content(
            model="gemini-2.5-flash-lite",
            contents=prompt,
            config=types.GenerateContentConfig(
                max_output_tokens=2000,
                temperature=temp,
            ),
        )
        print(run + 1, resp.text.strip() if resp.text else None)
```
